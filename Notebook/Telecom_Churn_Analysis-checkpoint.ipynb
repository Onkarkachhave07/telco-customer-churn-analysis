# 3. Exploratory Data Analysis (EDA)

```python
# What percentage of customers have churned?

plt.figure(figsize=(6,5))

churn_percentage = df["Churn"].value_counts(normalize=True) * 100

plt.pie(
    churn_percentage.values,
    labels=churn_percentage.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2", 2)
)

plt.title("Customer Churn Distribution")
plt.tight_layout()
plt.show()
```

## Out of all telecom customers, 73.46% remained with the company while 26.54% churned. This means approximately 1 in every 4 customers left the service, indicating customer churn is a significant business issue that requires further analysis.

```python
# Which contract type has the highest customer churn rate?

contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
)["Yes"] * 100

contract_churn = contract_churn.reset_index(name="Churn Rate (%)")

plt.figure(figsize=(8,5))

ax = sns.barplot(
    data=contract_churn,
    x="Contract",
    y="Churn Rate (%)",
    hue = "Contract",
    legend=False,
    palette="Set2"
)

for bar in ax.containers:
    ax.bar_label(bar, fmt="%.2f%%")

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.tight_layout()
plt.show()
```

## Customers with Month-to-Month contracts have the highest churn rate (42.71%), while customers with One-Year (11.27%) and Two-Year (2.83%) contracts have substantially lower churn rates. Churn decreases significantly as contract duration increases.

```python
# Does customer tenure influence churn?

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Churn",
    y="tenure",
    hue="Churn",
    palette="Set2",
    legend=False
)

plt.title("Customer Tenure by Churn Status")
plt.xlabel("Customer Churn")
plt.ylabel("Tenure (Months)")
plt.tight_layout()
plt.show()
```

## Customers who churn have a noticeably lower median tenure than customers who remain with the company. Most churned customers are concentrated in the early months of their subscription, while retained customers tend to have much longer tenures.

```python
# Do customers with higher monthly charges churn more?

plt.figure(figsize=(8,5))

sns.violinplot(
    data=df,
    x="Churn",
    y="MonthlyCharges",
    hue="Churn",
    palette="Set2",
    legend=False
)

plt.title("Monthly Charges by Churn Status")
plt.xlabel("Customer Churn")
plt.ylabel("Monthly Charges ($)")
plt.tight_layout()
plt.show()
```

## Customers who churn are concentrated in the higher monthly charge range, while retained customers are spread across both low and high monthly charge ranges. This suggests that customers paying higher monthly subscription charges are more likely to leave the service.

```python
# Which payment method has the highest churn rate?

payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
)["Yes"] * 100

payment_churn = payment_churn.reset_index(name="Churn Rate (%)")

plt.figure(figsize=(8,5))

ax = sns.barplot(
    data=payment_churn,
    x="PaymentMethod",
    y="Churn Rate (%)",
    hue="PaymentMethod",
    palette="Set2",
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f%%")

plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```

## Customers using Electronic Check have the highest churn rate among all payment methods. In contrast, customers using Bank Transfer (Automatic) and Credit Card (Automatic) show considerably lower churn rates, indicating better customer retention among automatic payment users.

```python
#  Which internet service has the highest churn rate?

internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
)["Yes"] * 100

internet_churn = internet_churn.reset_index(name="Churn Rate (%)")

plt.figure(figsize=(8,5))

ax = sns.barplot(
    data=internet_churn,
    x="InternetService",
    y="Churn Rate (%)",
    hue="InternetService",
    palette="Set2",
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f%%")

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.tight_layout()
plt.show()
```

## Customers using Fiber Optic internet service have the highest churn rate among all internet service categories. Customers with DSL show a considerably lower churn rate, while customers without internet service exhibit the lowest churn rate.

```python
# What is the revenue impact of customer churn based on average monthly charges?

plt.figure(figsize=(8,5))

ax = sns.barplot(
    data=df,
    x="Churn",
    y="MonthlyCharges",
    hue="Churn",
    palette="Set2",
    legend=False,
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(container, fmt="$%.2f")

plt.title("Average Monthly Charges by Churn Status")
plt.xlabel("Customer Churn")
plt.ylabel("Average Monthly Charges ($)")
plt.tight_layout()
plt.show()
```

## Customers who churn have a higher average monthly charge than retained customers. This indicates that customer churn does not only affect customer count but also has a direct impact on recurring monthly revenue.

```python
# Which combination of contract type and internet service has the highest churn risk?

segment_churn = pd.crosstab(
    [df["Contract"], df["InternetService"]],
    df["Churn"],
    normalize="index"
)["Yes"] * 100

segment_churn = segment_churn.unstack()

plt.figure(figsize=(8,5))

sns.heatmap(
    segment_churn,
    annot=True,
    fmt=".2f",
    cmap="Reds",
    linewidths=0.5,
    cbar_kws={"label": "Churn Rate (%)"}
)

plt.title("Churn Rate by Contract Type and Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Contract Type")
plt.tight_layout()
plt.show()
```

## The highest churn rate (54.61%) is observed among customers with Month-to-Month contracts and Fiber Optic internet service, making this the highest-risk customer segment. Churn decreases consistently as contract duration increases across all internet service categories, with Two-Year contract customers showing the lowest churn rates regardless of internet service type.

```python
## Are senior citizens more likely to churn?

senior_citizen = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
)["Yes"] * 100

senior_citizen = senior_citizen.reset_index(name="Churn Rate (%)")

plt.figure(figsize=(8,5))

ax = sns.barplot(
    data=senior_citizen,
    x="SeniorCitizen",
    y="Churn Rate (%)",
    hue="SeniorCitizen",
    palette="Set2",
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f%%")

plt.title("Churn Rate by Senior Citizen")
plt.xlabel("Senior Citizen")
plt.ylabel("Churn Rate (%)")
plt.tight_layout()
plt.show()
```

## Senior citizens have a 41.68% churn rate, compared to 23.61% for non-senior citizens. This indicates that senior citizens are considerably more likely to leave the telecom service than younger customers.
